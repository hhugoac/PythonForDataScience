```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
np.set_printoptions(precision=2)
```

```python
#Load data set
college = pd.read_csv('College.csv')
college
```

```python
college2 = pd.read_csv('College.csv', index_col=0)
college2
```

```python
college_updated = college.rename({'Unnamed: 0':'College'}, axis=1)
college_updated = college_updated.set_index('College')
college_updated
```

```python
college = college_updated
college
```

```python
college.describe()
```

```python
columns = ['Top10perc', 'Apps', 'Enroll']
pd.plotting.scatter_matrix(college[columns])
```

```python
fig, axes = plt.subplots(ncols=2, figsize=(12,12))
college.boxplot('Outstate', by = 'Private', ax = axes[0])
```

```python
college['Elite'] = pd.cut(college['Top10perc'], [0, 50, 100], labels=['No', 'Yes'])
```

```python
college['Elite'].value_counts()
```

```python
fig, axes = plt.subplots(ncols=3, figsize=(12,12))
college.plot.hist(column='Grad.Rate', by = 'Elite', ax = axes[2])
```

```python
college['Acceptance_perc'] = college['Accept'] / college['Apps'] * 100
```

```python
college['Acceptance_perc'].describe()
```
