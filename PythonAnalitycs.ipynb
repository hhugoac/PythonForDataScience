```python
#Sample Monte-Carlo simulation
import numpy as np

# Parameter definitions
n_simulation = 10000
results = []

# Monte-Carlo simulation
for _ in range(n_simulation):
    #Simulate a round
    if np.random.rand() > 0.5:
        results.append(1)  # Win 1
    else:
        results.append(-1) # Lose 1

# Estimate the expected value (mean)
expected_value = np.mean(results)
print(f'Expected value: {expected_value}')
```

```python
# functional programming
results_functional = np.mean(
    [ 1 if np.random.rand() > 0.5 else -1 for _ in range(n_simulation) ]
)
print(f'Expected value: {results_functional}')
print(results_functional)
```
