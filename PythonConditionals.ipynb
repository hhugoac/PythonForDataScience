# IF, IFEL, ELSE

#### Simple if and else: Determinate if a number is even or odd

```python
number = int(input("Type a number: "))

if number % 2  == 0:
    print("even")
else:
    print("odd")
```

Multiple if: estimate the price according the age

```python
age = int(input("Type your age"))
if age < 12:
    price = 50
elif age < 18:
    price = 80
else:
    price = 100
print(f"The ticket costs {price} pesos")
```

#### Use of if to validate information

```python
user = "admin"
password = input("Type the password: ")
if user == "admin" and password == "1234":
    print("Access granted")
else:
    print("Access denied")
```

#### Nested if, select the student category accordingly the score

```python
score = int(input("Type your score: "))

if score >=90:
    if score >= 95:
        print("Excellent! You got A+")
    else:
        print("You got A")
elif score >= 80:
    print("Excellent! You got B")
else:
    print("You need improve, you got C or less.")
```

#### Combine if, elif and else

```python
weight = float(input("Type your weight: "))
height = float(input("Type your height: "))
imc = weight / (height ** 2)

if imc < 18.5:
    print(f"Underweight {round(imc, 2)}")
elif imc < 24.9:
    print(f"Normal {round(imc, 2)}")
elif imc < 29.9:
    print(f"Overweight {round(imc, 2)}")
else:
    print(f"Obesity {round(imc, 2)}")
```

#### Conditionals operators and try/except

```python
a = int(input("Type a number: "))
b = int(input("Type another number: "))
try:
    result = a / b
except ZeroDivisionError:
    print("Error: Division by zero")
    result = None
except TypeError:
    print("Error: Type a number")
    result = None
else:
    if result > 0:
        print("The result is positive")
    elif result < 0:
        print("The result is negative")
    else:
        print("The result is zero")
finally:
    print("Running try/except block finishes")
```
