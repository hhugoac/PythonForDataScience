```python
# Recursivity, calculate the factorial of n

def recursivity_factorial(n):
    if n <= 1:
        return 1
    else:
        return n * recursivity_factorial(n-1)
number_to_factorial = int(input("Type a number to calculate its factorial: "))
print(f'The factorial of {number_to_factorial} is {recursivity_factorial(number_to_factorial)}')
```

```python
# Recursivity, fibonacci of a number n
def recursivity_fibonacci(n):
    if n <= 0:
        print(f'the value of n is {n}')
        return 0
    elif n == 1:
        print(f'the value of n is {n}')
        return 1
    print(f'the value of n is {n}')
    return recursivity_fibonacci(n-1) + recursivity_fibonacci(n-2)

number_to_fibonacci = int(input("Type a number to calculate its fibonacci: "))
print()
print(f'The fibonacci of {number_to_fibonacci} is {recursivity_fibonacci(number_to_fibonacci)}')
```
