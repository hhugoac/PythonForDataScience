# Variables and operators

### Definición de variables

```
# Operations sample

salary_base = 1500 # Salary base
perk_productivity= 3000 # Productivity perk
tax_percentage = 12 # Tax percentage
```

### Operations

```
# Gross salary
salary_gross = salary_base + perk_productivity

# Tax
tax = salary_gross * tax_percentage / 100

# Salary after tax
salary_after_tax = salary_gross - tax
```

### Impresión de resultados en pantalla

```
print("Salary Monthly")
print("Salary gross $",salary_gross, " pesos")
print("Productivity perk $", perk_productivity, " pesos")
print("Tax (12%) $", tax_percentage, " pesos")
print("Salary Net $", salary_after_tax, " pesos")
```

# Working with text strings

## Concatenation

```
## With operator +
name = "Hugo"
another_name = "Hector"
full_name = another_name + " " + name
print(full_name) # Result
```

```
## With format strings cleaner
f_full_name = f"My first name is {another_name} and my middle name is {name}"
print(f_full_name)
```

```
## With join methos (for lists)
list_of_words = ["Python", "is", "a", "language"]
join_name =" ".join(list_of_words)
print(join_name)
```

### Extracting characters

```
my_variable_for_extracting = "Hugo"
print(my_variable_for_extracting[0])
```

### Slicing (substrings)

```
my_slicing_string = "This a text to slice it"
print(my_slicing_string[0:4]) ### this prints the first chars
print(my_slicing_string[0:]) ### this prints the first chars
print(my_slicing_string[0:6]) ### this prints the first chars
```

### Length of the string

```
my_variable_calculate_length = "This a text has a length unknown of chars"
print(len(my_variable_calculate_length))
```

### Upper case and lower case

```
my_var_to_upper_case = "asdfghjkl"
print(my_var_to_upper_case.upper())

my_var_to_lower_case = "AfgdRT_ asdERTT"
print(my_var_to_lower_case.lower())
```

### Replacement of substrings

```
original_text = "In this text would be replaced chars."
replaced_char_e_with_o = original_text.replace("e", "o")
replaced_char_space_with_undercore = original_text.replace(" ", "_")
print(replaced_char_e_with_o)
print(replaced_char_space_with_undercore)
```

### Deleting white space

```
text_with_left_and_rigth_spaces = " This text contains both left space and right "
print(text_with_left_and_rigth_spaces.rstrip())
print(len(text_with_left_and_rigth_spaces.rstrip()))
print(text_with_left_and_rigth_spaces.lstrip())
print(len(text_with_left_and_rigth_spaces.lstrip()))
print(text_with_left_and_rigth_spaces.strip())
print(len(text_with_left_and_rigth_spaces.strip()))
```

### Lists

```
shopping_list = ["bread", "eggs", "milk", "fruit"]
```

#### Add items to the list

```
shopping_list.append("rice")
shopping_list.append("chicken")

print(shopping_list)
```

#### Delete an item from the list

```
shopping_list.remove("bread")
print(shopping_list)
```

#### Sort the list

```
shopping_list.sort()
print(shopping_list)
```

#### Show the list

```
print("Shopping list updated:")
print(shopping_list)
```

#### Show first and last elements

```
print("First elements: ", shopping_list[:2])
print("Last elements: ", shopping_list[-2:])
```

### Tuples

```
city1 = ("Mexico City", 19.4326, -99.1332)
city2 = ("Guadalajara", 20.6597, -103.3496)
city3 = ("Monterrey", 25.6866, -100.3161)
```

#### Tuples operations

```
total_longs = city1[1] + city2[1]
total_lats = city1[2] + city2[2]

new_tuple = (total_longs, total_lats)
```

#### Results

```
print("Cities coordinates")
print("City 1", city1)
print("City 2", city2)
print("City 3", city3)
print("Sum of coordinates with two cities: ", new_tuple)
```
