```python
# Complex numbers class

class ComplexNumbers:

    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def add(self, number):
        real = self.real + number.real
        imaginary = self.imaginary + number.real
        result = ComplexNumbers(real, imaginary)
        return result
```

```python
complex_1 = ComplexNumbers(3,4)
complex_2 = ComplexNumbers(2,8)

result_sum = complex_1.add(complex_2)
print(f'Result = Real: {complex_1.real}, Imaginary: {complex_1.imaginary} = {result_sum.real} + {complex_1.imaginary}i')
```

```python
class Person:
    def __init__(object, name, age):
        object.name = name
        object.age = age

    def message(object):
        print(f'Mu name is {object.name} and I am {object.age} years old')

hugo = Person('Hugo', 36)
hugo.message()
```

```python
class Triangle:

    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def perimeter(self):
        return self.a + self.b + self.c
```

```python
t1 = Triangle(1, 2, 3)
print(f'The perimeter of triangle one is {t1.perimeter()} ')
```

```python
# Class to handle a bank account
class BankAccount:
    def __init__(self, name, initial_balance):
        self.name = name
        self.balance = initial_balance

    def deposit(self, amount):
        self.balance += amount
        print(f'You deposit {amount} from this account, your balance is {self.balance}')


    def withdraw(self, amount):
        if amount > self.balance:
            print(f'You cannot withdraw {amount} from this account')
        else:
            self.balance -= amount
            print(f'You withdraw {amount} from this account, your balance is {self.balance}')

my_account = BankAccount('Hugo', 1000)
my_account.deposit(100)
my_account.withdraw(110)
my_account.withdraw(90)
```

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def start(self):
        print(f'A vehicle of brand  {self.brand} with model{self.model} is starting.')


class Truck(Vehicle):
    def __init__(self, brand, model, doors):
        super().__init__(brand, model)
        self.doors = doors


    def play_beep(self):
        print('Beep beep!')

my_truck = Truck('Toyota', 'Corolla', 5)
my_truck.start()
my_truck.play_beep()

```

```python
# Class with getters and setter
class Product:

    def __init__(self, name, price):
        self.name = name
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price > 0:
            self.__price = price
        else:
            print('Sorry, the price cannot be less than 0')

my_product = Product('computer', 30000)
my_product.set_price(-100)
my_product.set_price(15000)
my_product.get_price()
```

```python
class Bird:
    def fly(self):
        print('Bird fly!')

class Penguin(Bird):
    def fly(self):
        print('Penguin cannot fly!')

my_bird = Bird()
my_bird.fly()

my_penguin = Penguin()
my_penguin.fly()
```

```python
class Book:
    def __init__(self, name, author):
        self.name = name
        self.author = author

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def show_books(self):
        print(f'Book in your library')
        [ print(f'Comprenhension Book name: {book.name}, author: {book.author}') for book in self.books]
        list(map(lambda book: print(f'Lambda Book name: {book.name}, author: {book.author}'), self.books))
        for book in self.books:
            print(f'Book name: {book.name}, author: {book.author}')

my_library = Library()
my_library.add_book(Book('El quijote de la Mancha', 'Cervantes'))
my_library.add_book(Book('Doña Barbara', 'Romulo Gallegos'))
my_library.show_books()
```
