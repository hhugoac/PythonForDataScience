# Variables y operadores

### Definición de variables

```
# Ejemplo de operaciones

salario_base = 1500 # Bono adicional de productividad
bono_productividad = 3000 # Bono adicional de productividad
descuento_impuesto = 12 # Porcentaje de impuestos
```

### Cálculo de operaciones

```
# Cálculo salario bruto
salario_bruto = salario_base + bono_productividad

# Cálculo de impuestos
impuestos = salario_bruto * descuento_impuesto / 100

# Cálculo salario neto
salario_neto = salario_bruto - impuestos
```

### Impresión de resultados en pantalla

```
print("Salario Mensual")
print("Salario base $",salario_neto, " pesos")
print("Bono productividad $", bono_productividad, " pesos")
print("Impuestos (12%) $",impuestos, " pesos")
print("Salario Neto $", salario_neto, " pesos")
```
