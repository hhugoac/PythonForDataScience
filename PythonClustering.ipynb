# Hirarchical Cluster

```python
# load libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load data as dd with pandas
df = pd.read_csv('WholesaleCustomersData.csv')
df.sample(5)
```

```python
# load normalization library
from sklearn.preprocessing import normalize
data_scaled = normalize(df)
data_scaled
```

```python
data_scaled = pd.DataFrame(data_scaled, columns=df.columns)
data_scaled.sample(5)
```

```python
# load library for hirachical clustering and dendrogram
import scipy.cluster.hierarchy as sch

plt.figure(figsize=(10, 7))
plt.title('Dendrogram')
dend = sch.dendrogram(sch.linkage(data_scaled, method='ward'))
```

```python
# Print number groups of dendrogram
print(dend['color_list'])
```

```python
# print the eunique colors
unique_colors = set(dend['color_list'])
unique_colors
```

```python
num_clusters_optimal = len(unique_colors) - 1
num_clusters_optimal
```

```python
plt.figure(figsize=(10, 7))
plt.title('Dendrogram')
dend = sch.dendrogram(sch.linkage(data_scaled, method= 'ward'))
plt.axhline(y = 6, color = 'r', linestyle = '--')
```

```python
# Aglomerative Clustering
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2, metric = 'euclidean', linkage= 'ward')
groups = cluster.fit_predict(data_scaled)
groups
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(data_scaled['Milk'], data_scaled['Grocery'], c = cluster.labels_)
```

```python
# Group segregation by PCA
campos = data_scaled.values
campos
```

```python
from sklearn import decomposition

pca = decomposition.PCA(n_components=2)
pca.fit(campos)
campos = pca.transform(campos)
campos
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(campos[:,0], campos[:,1], c = cluster.labels_)
```

```python
data_frame = pd.DataFrame(groups, columns=['Group'])
data_frame
```

```python
data_frame2 = pd.concat([df, data_frame], axis = 1, join = 'inner')
data_frame2
```

```python
pd.options.display.max_rows = None
data_frame2
```
