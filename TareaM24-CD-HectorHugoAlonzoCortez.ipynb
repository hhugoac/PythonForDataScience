```python
# load libraries 
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
# load data set from drugs.csv file
drugs_data_frame = pd.read_csv('drugs.csv')
drugs_data_frame.sample(5)
```

```python
# set the columns to used as feature matrix X and target variable Y
feature_names = drugs_data_frame.columns.to_list()
feature_names.remove('Drug')

X = drugs_data_frame[feature_names].values
y = drugs_data_frame['Drug']
```

```python
# Preprocessing to generate numerical values from categorical values
from sklearn import preprocessing

sex_codfication = preprocessing.LabelEncoder()
sex_codfication.fit(['F', 'M'])
X[:, 1] = sex_codfication.transform(X[:, 1])

bp_status_codfication = preprocessing.LabelEncoder()
bp_status_codfication.fit(['HIGH', 'LOW', 'NORMAL'])
X[:, 2] = bp_status_codfication.transform(X[:, 2])

cholesterol_status_codfication = preprocessing.LabelEncoder()
cholesterol_status_codfication.fit(['HIGH', 'NORMAL'])
X[:, 3] = bp_status_codfication.transform(X[:, 3])
```

```python
# Build the trainig and testing sets 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

## Función para evaluar un modelo

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import classification_report

def evaluate_model(model):
    model.fit(X_train, y_train)
    print(f'Score model for data trained:  {model.score(X_train, y_train)}')
    print(f'Score model for data test: {model.score(X_test, y_test)}')
    y_hat = model.predict(X_test)
    print(classification_report(y_test, y_hat))
```

## Random forest

```python
# load libraries for random forest
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100, random_state=1)

evaluate_model(rf)
```

## Gradient boost

```python
# load gradient boosting claisifier library
from sklearn.ensemble import GradientBoostingClassifier
gbt = GradientBoostingClassifier(n_estimators=100, random_state=1) # 100 levels of depth, 3 is the default
evaluate_model(gbt)

gbt2 = GradientBoostingClassifier(n_estimators=100, random_state=1, max_depth=2)
evaluate_model(gbt2) # 100 levels of depth, 3 is the default
```

### AdaBoost classifier

```python
from sklearn.ensemble import AdaBoostClassifier
abc = AdaBoostClassifier(n_estimators=50, learning_rate=1.0)
evaluate_model(abc)
```

```python
classifier_svm = CalibratedClassifierCV(SVC(kernel='linear'), ensemble=False) 
ada_svm_linear = AdaBoostClassifier(n_estimators=50, estimator=classifier_svm, learning_rate=1.0)
evaluate_model(ada_svm_linear)
```
