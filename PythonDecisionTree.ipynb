# Python Decision Tree

```python
# load libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
import matplotlib.pyplot as plt
 
```

## Sample 1

```python
# load fraud dataset
fraud_data = pd.read_csv('Fraude.csv')
fraud_data.sample(5)
```

```python
# Set the columns to be used as features matrix X and target variable Y
feature_columns = ['Reembolso', 'EdoCivil', 'Ingresos']
X = fraud_data[feature_columns].values
Y = fraud_data.Fraude
```

```python
# Preprocessing to generate numerical values from categorical values
from sklearn import preprocessing

refound_codfication = preprocessing.LabelEncoder()
refound_codfication.fit(['No', 'Sí'])
X[:, 0] = refound_codfication.transform(X[:, 0])

marital_status_codfication = preprocessing.LabelEncoder()
marital_status_codfication.fit(['Casado', 'Divorciado', 'Soltero'])
X[:, 1] = marital_status_codfication.transform(X[:, 1])
```

```python
X[0:5]
```

```python
# Build training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=1)
```

```python
# Build the decision tree classifier model
clf = DecisionTreeClassifier()

# Apply the model to the training data
clf = clf.fit(X_train, Y_train)

# Predict the response for the test dataset
y_hat = clf.predict(X_test)
```

```python
X_test
```

```python
y_hat
```

```python
Y_test
```

```python
# Confusion matrix
confusion_matrix = metrics.confusion_matrix(Y_test, y_hat)
confusion_matrix
```

```python
# Show the report of the model
print(metrics.classification_report(Y_test, y_hat))
```

```python
# load more libraries 
from sklearn import tree
import graphviz
import pydotplus
from PIL import Image
```

```python
# Set the decision rules to be used in the decision tree
dot_data = tree.export_graphviz(clf, out_file=None, 
                                feature_names=feature_columns,  
                                class_names=['No', 'Sí'])

# Build the decision tree graph
graph = pydotplus.graph_from_dot_data(dot_data)
```

```python
# Export the decision tree graph to a PDF file
graph.write_pdf("DecisionTree.pdf")

# Export the decision tree graph to a PNG file
graph.write_png("DecisionTree.png")

image = Image.open("DecisionTree.png")
image .show()
```
