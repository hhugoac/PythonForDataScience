# Custom Seaborn plots

- Use the kaggle data set
- Make a loop with different styles to change the hist style
- Remove axis with despite
- Combine matplotlib with seaborn
- Customize the color palette

```python
# Use the kaggle data set
# Make a loop with different styles to change the hist style
for style in ['dark', 'whitegrid', 'darkgrid', 'ticks']:
    sns.set_style(style)
    sns.displot(df['reading score'])
```

```python
# Remove axis with despite

sns.set_style('ticks')
sns.displot(df['reading score'], kde=True)
sns.despine(left=True, bottom=True)
```

```python
# Combine matplotlib with seaborn
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
sns.distplot(df['writing score'], ax = ax)     # Deprecated version
ax.set(xlabel='Writing Score', ylabel='Reading Score')

# Customize the color palette
```

```python
fig, ax = plt.subplots()
sns.histplot(df['writing score'], ax = ax, kde= True)    # Updated version
ax.set(xlabel='Writing Score', ylabel='Reading Score')
```

```python
# Trunk the visualization scale
fig, ax  = plt.subplots()
sns.histplot(df['writing score'], ax = ax, kde= True)
ax.set(xlabel='Writing Score', ylabel='Reading Score', xlim = (0, 90), title= 'Frecuency Histogram for writing score')
# Customize the color palette
```

```python
fig, (ax0, ax1) = plt.subplots(nrows=1, ncols=2, sharey=True)
sns.histplot(df['writing score'], ax = ax0, kde= True) # Histogram for all observations
sns.histplot(df.query('gender == "male"')['writing score'], ax = ax1, kde= True)  # Histogram for men
ax0.set(xlabel='General Writing Score')
ax1.set(xlabel='Male Writing Score')
ax1.axvline(x = 75, label = 'Archivement',  linestyle = '--')
ax1.legend()
```

```python
palettes = ['deep', 'muted', 'pastel', 'bright', 'dark']

for palette in palettes:
    sns.set_palette(palette)
    sns.displot(df['reading score'], kde= True)
```

```python
def palette_plot(palette_input):
    palettes = sns.color_palette(palette_input, 20)

    fig, axes = plt.subplots(figsize=(8, 2))

    for idx, palette in enumerate(palettes):
        axes.add_patch(plt.Rectangle((idx, 0), 1, 1, color=palette))

    axes.set_title(palette_input)
    axes.set_xlim(0, len(palettes))
    axes.set_ylim(0, 1)
    axes.set_axis_off()
    plt.show()
```

```python
# Customize the color palette
palette_plot('deep')
palette_plot('coolwarm')
palette_plot('pastel')
palette_plot('bright')
palette_plot('dark')
palette_plot('colorblind')
```

# Compound plots

**Goal for the lesson**

- Use of facetgrid

```python
# Use of facetgrid
graph = sns.FacetGrid(df, col = 'race/ethnicity', col_order=['group A', 'group B', 'group C', 'group E', 'group E'])
graph.map(sns.boxplot, 'math score')
plt.show()
```

```python
# factor plot
sns.catplot(x = 'math score', data = df, col = 'race/ethnicity', kind = 'box')
```

```python
# dispersion plots
graph = sns.FacetGrid(df, col = 'race/ethnicity')
graph.map(sns.scatterplot, 'reading score', 'writing score')
```

```python
# regression tabular plots lmplot
sns.lmplot(data = df, x = 'reading score', y = 'writing score', col = 'gender', row = 'race/ethnicity' )
```

```python
# jointGrid, regression and dispersion
graph = sns.JointGrid(df, x = 'reading score', y = 'writing score')
graph.plot(sns.regplot, sns.histplot)
# jointplot kind = 'scatter'
```

```python
sns.jointplot(data = df, x = 'reading score', y = 'writing score', kind = 'hex')
```

```python
# Scattering plots with categorical variables
# load the data
tips = sns.load_dataset('tips')
```

```python
# Build scatter plot
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'time', style = 'sex', palette = 'Set2')
plt.title('Total account vs. tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Box plot
# Build a box plot to analyze variability
sns.boxplot(data = tips, x = 'day', y = 'total_bill', hue = 'sex', palette = 'pastel')
plt.title('Account distribution by day and gender')
plt.xlabel('Day')
plt.ylabel('Total account')
plt.legend(title='Gender')
plt.show()
```

```python
# Heat map to analyze correlation
# Estimate the correlation matrix
tips_numerical_values = tips.select_dtypes(include = 'number')
correlation_matrix = tips_numerical_values.corr()
```

```python
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Pair title for relations between multiple variables
# Build pair plot to analyze relations between columns
sns.pairplot(data = tips, hue = 'time', palette = 'husl', diag_kind = 'kde')
plt.suptitle('Relations in the variables in the dataset Tips', y = 1.02)
plt.show()
```

```python
# Violin plot with categorical and numerical plots
sns.violinplot(data = tips, x = 'day', y = 'total_bill', hue = 'time', split= True,  palette = 'muted')
plt.title('Distribution of day accounts and eat time')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# facet grid for composed plots
# Build a facetgrid to analyze tip by gender and time
g = sns.FacetGrid(data = tips, col = 'sex', row =  'time', hue = 'day', palette = 'bright', margin_titles=True)
g.map(sns.scatterplot, 'total_bill', 'tip', alpha = 0.7)
g.add_legend()
g.fig.suptitle('Tips by account, gender, and time', y = 1.05)
plt.show()
```
