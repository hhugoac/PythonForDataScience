```python
import pandas as pd
import seaborn as sns
```

```python
# Download and load data
df = pd.read_csv('StudentsPerformance.csv')
df
```

### Comparison between the pandas and matplotlib with seaborn

```python
df['math score'].plot.hist()
```

```python
# Use of distplot from seaborn

sns.displot(df['math score'])
```

```python
sns.distplot(df['math score']) # deprecated
```

```python
#Use of displot, kde, bins

sns.histplot(df['math score'], kde=True) # Updated version
```

```python
#Use of displot hist false
sns.distplot(df['math score'], hist=False)
```

```python
sns.distplot(df['math score'], hist=False, kde_kws={'shade': True})
```

```python
# Scatter plots in seaborn with regplot
sns.regplot(x='reading score', y='writing score', data=df )
```

```python
sns.lmplot(x='reading score', y='writing score', hue='gender', data=df)
```

```python
sns.lmplot(x='reading score', y='writing score', col='gender', data=df)
```

```python
sns.lmplot(x='reading score', y='writing score', data=df, col = 'race/ethnicity')
```
