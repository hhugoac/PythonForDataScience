```python
# EDA with descriptive statistical
my_data_3 = {
"Producto": ["Manzanas", "Bananas", "Naranjas", "Uvas", "Peras"],
"Cantidad": [10, 20, 15, 12, 8],
"Precio": [5.5, 2.5, 3.0, 4.5, 6.0],
"Categoria": ["Frutas", "Frutas", "Frutas", "Frutas", "Frutas"]}
```

```python
my_dp= pd.DataFrame(my_data_3)
```

```python
print('Dataframe EDA')
print(my_dp.describe())
```

```python
mean_quantity = my_dp['Cantidad'].mean()
sum_quantity = my_dp['Cantidad'].sum()

print(f'Mean quantity: {mean_quantity}')
print(f'Sum quantity: {sum_quantity}')
```

### Filtering multiple conditions

```python
# Filter products with a quantity great than 10 and less than 5
filter_1 = my_dp['Cantidad'] > 10
filter_2 = my_dp['Precio'] < 5

condition_3 = my_dp[filter_1 & filter_2]
print('Products with price less than 5 and quantity great than 10')
print(condition_3)
```

### Pivot table for data analisys

```python
datos_ext = {
"Producto": ["Manzanas", "Bananas", "Naranjas", "Uvas", "Manzanas", "Bananas"],
"Cantidad": [10, 20, 15, 12, 5, 10],
"Precio": [5.5, 2.5, 3.0, 4.5, 5.5, 2.5],
"Mes": ["Enero", "Enero", "Enero", "Enero", "Febrero", "Febrero"]
}

my_dp_ext = pd.DataFrame(datos_ext)
```

```python
# Create a pivot table
pivot_table = my_dp_ext.pivot_table(values= 'Cantidad', index= 'Producto', columns='Precio', aggfunc= 'sum', fill_value=0)
print('Dynamic table of quantities by month')
print(pivot_table)
```

```python
my_dp_ext.describe()
print(my_dp_ext)
```

### Custom functions

```python
# Classify with price
def classify_by_price(price):
    if price < 3:
        return 'Cheap'
    elif price < 5:
        return 'Reasonable'
    else:
        return 'Expensive'
```

```python
# Apply a new colum this new classification
my_dp_ext['Price classification'] = my_dp_ext['Precio'].apply(classify_by_price)
print('Data table with price classification')
print(my_dp_ext)
```

### Operations with duplicated data


```python
new_row = pd.DataFrame({"Producto": ["Manzanas"], "Cantidad": [10], "Precio": [5.5], "Categoria": ["Frutas"], "Price classification": [classify_by_price(5.5)]})

my_data_ext = pd.concat([new_row, my_dp_ext], ignore_index=True)
print(my_data_ext)

```
