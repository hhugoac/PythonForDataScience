# Matplotlib

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create the initial plots
fig, ax = plt.subplots()
plt.show()
```

```python
# Dummy data frames
months = pd.DataFrame(np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']), columns = ['Months'])
temperatures = pd.DataFrame(np.array([20.3, 22.5, 18.6, 21.2, 23.3, 23.6, 26.4, 28.5, 25.1, 23.3, 22.1, 21.9]), columns = ['Temperatures'])
```

```python
# Plot temperature vs months
fig, ax = plt.subplots()
ax.plot(months['Months'], temperatures['Temperatures'])
plt.show()
```

```python
# Temperatures X for a city
temperatures_X = pd.DataFrame(np.array([17.3, 20.5, 20.6, 24.2, 26.3, 24.6, 28.4, 30.5, 26.1, 24.3, 24.1, 20.9]), columns = ['Temperatures'])
fig, ax = plt.subplots()
ax.plot(months['Months'], temperatures['Temperatures'])
ax.plot(months['Months'], temperatures_X['Temperatures'])
plt.show()
```

```python
# Plot with labels
fig, ax = plt.subplots()
ax.plot(months['Months'], temperatures['Temperatures'])
ax.plot(months['Months'], temperatures_X['Temperatures'])
ax.set_xlabel('Months')
ax.set_ylabel('Temperatures')
plt.show()
```

```python
# Plot with labels and custom colors for the lines
fig, ax = plt.subplots()
ax.plot(months['Months'], temperatures['Temperatures'], color='red')
ax.plot(months['Months'], temperatures_X['Temperatures'], color='blue')
ax.set_xlabel('Months')
ax.set_ylabel('Temperatures')
plt.show()
```

```python
# Subplots for each temperature
fig, ax = plt.subplots(2, 1)
ax[0].plot(months['Months'], temperatures['Temperatures'], color='magenta')
ax[1].plot(months['Months'], temperatures_X['Temperatures'], color='pink')
ax[1].set_xlabel('Months')
ax[0].set_ylabel('Temperatures')
ax[1].set_ylabel('Temperatures for X')
plt.show()
```

```python
# Subplots for each temperature
fig, ax = plt.subplots(3, 2)
ax[0,0].plot(months['Months'], temperatures['Temperatures'], color='magenta')
ax[0,1].plot(months['Months'], temperatures_X['Temperatures'], color='pink')
ax[0,0].set_ylabel('Temperatures')
plt.show()
```

```python
# Create a time serie data function
def plot_time_series(axis, x, y, color, x_label, y_label):
    axis.plot(x, y, color=color)
    axis.set_xlabel(x_label)
    axis.set_ylabel(y_label, color=color)
    axis.tick_params('y', colors=color)
```

```python
# Create plot with customized colors
fig, ax = plt.subplots()
x_label = 'Monthly data: 2015-01-01 - 2018-09-01'
y_label = 'CO2 emissions'
plot_time_series(ax, filtered_data_by_date['Date'], filtered_data_by_date['Interpolated'], 'blue', x_label, y_label)

ax.axes.xaxis.set_ticks([])
plt.show()
```

```python
# Create a trend vs date plot with ax.twinx()
fig, ax = plt.subplots()
x_label = 'Monthly data: 2015-01-01 - 2018-09-01'
y_label = 'CO2 emissions'
plot_time_series(ax, filtered_data_by_date['Date'], filtered_data_by_date['Interpolated'], 'blue', x_label, y_label)

ax_2 = ax.twinx()
plot_time_series(ax_2, filtered_data_by_date['Date'], filtered_data_by_date['Trend'], 'red', '', y_label + ' trend')

ax.axes.xaxis.set_ticks([])
plt.show()
```
