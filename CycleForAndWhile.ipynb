# Cycle While and For

```python
# Count until 5
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# Dynamic condition. Ask password until true

password = input("Type password: ")

while True:
    if password == "python123":
        print("Access granted")
        break
    else:
        print("Access denied")
        password = input("Type password: ")
        continue
```

```python
# Dynamic condition. Ask password until true (solution from class)
password = "python123"
input_text = ""

while input_text != password:
    input_text = input("Type your password.")
print("Successful password!!")
```

```python
# Show the elements of a list of fruits
fruits = ["apple", "mango", "grape", "banana"]
# With for
print("\nWith for cycle")
for fruit in fruits:
    print(fruit)

#With while
print("\nWith while cycle")
count = 0
while count < len(fruits):
    print(fruits[count])
    count += 1
```

```python
# Cycle for with range
for i in range(1,11,1):
    print(i)
```

```python
# Iteration of a dictionary with for
dict_capitals = {"Germany": "Berlin", "Spain": "Madrid", "Argentina": "Buenos Aires", "Italy": "Rome", "China": "Pekin"}

for country, capital in dict_capitals.items():
    print(f"The capital of {country} is {capital}")
```

```python
# Times table
number = int(input("Type a number to calculate the table: "))

for i in range(1,11,1):
    print(f" {number} x {i} = {i * number} ")
```

```python
# While cycle: estimate the sum until the user wants

sum = 0

while True:
    number = int(input("Type a number to add to the sum: "))
    sum += number
    question = input("Do you want to add another number? (Y/N): ")
    if question == "Y":
        continue
    else:
        print(f"The total is {sum}")
        break
```

```python
# While cycle: estimate the sum until the user wants (official solution)
sum = 0
is_continuing = "Y"

while is_continuing == "Y":
    number = int(input("Type a number to add to the total: "))
    sum += number
    is_continuing = input("Do you want to add another number? (Y/N): ")
print(f"Sum is: {sum} ")
```

```python
# Nested for print the matriz index

for i in range(1,4,1):
    for j in range(1,4, 1):
        print(i,j)
```

# Intro to functional programming paradigm

```python
# Modify data in a list
numbers = [1, 2, 3, 4, 5]
numbers_modified = [x * 2 for x in numbers]
print(numbers_modified)
```

```python
# Filter data in a list
number_filtered = [x for x in numbers if x > 2]
print(number_filtered)
```

```python
# Modify and filter data in a list
numbers_modified_and_filtered = [x * 2 for x in numbers if x > 2]
print(numbers_modified_and_filtered)
```

```python
# Working with strings

strings = ["one", "two", "three", "four", "five"]
strings_upper = [ word.upper() for word in strings if len(word) > 3]
print(strings_upper)
```

```python
# Filter with multiple conditions, get the even numbers greater than 2

long_list = range(1,10,1)
even_gt_two = [ x for x in long_list if x % 2 == 0 and x > 2]
odd_gt_two = [ x for x in long_list if x % 2 != 0 and x > 2]
print(even_gt_two)
print(odd_gt_two)
```

## Sample of while and for

```python
numbers_list_for = range(1, 5, 1)
sum = 0
for number in numbers_list_for:
    sum += number
print(f"Sum is {sum}")

# Fibonacci sequence until 50 with while
a, b = 0, 1
while a <= 50:
    print(a, end = ' ')
    a, b = b, b + a
print()
```
